```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data['Education']=le.fit_transform(data['Education'])
data['Property_Area']=le.fit_transform(data['Property_Area'])
data['Loan_Status']=le.fit_transform(data['Loan_Status'])
data.head(10)
```

```python
from sklearn.model_selection import train_test_split
x=data.iloc[:,[4,6,7,8,9,10,11]]
y=data.iloc[:,12]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
rf.fit(x_train,y_train)
print("Here 0 means no and 1 means yes!")
y_pred=rf.predict(x_test)
y_pred
```

```python
education=int(input("Enter your education(graduate=0,not gradudate=1): "))
applicantincome=int(input("Enter the applicant income: "))
coapplicantincome=int(input("Enter the coapplicant income: "))
loanamount=int(input("Enter your loan amount: "))
loanamountterm=int(input("Enter your loan amount term: "))
credithistory=int(input("Enter your credit history(0-1): "))
propertyarea=int(input("Enter your property area(0-2): "))
new_data=[[education,applicantincome,coapplicantincome,loanamount,loanamountterm,credithistory,propertyarea]]
new_data_scaled=sc.transform(new_data)
prediction=rf.predict(new_data_scaled)
print("Your prediction is: ",prediction)
if(prediction==0): 
    print("I am sorry to say, You are not eligible for loan!")
else:
    print("Congratulations!, You are eligible for loan")
```

```python
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,roc_auc_score,confusion_matrix)
acc=accuracy_score(y_test,y_pred)
print("accuracy score = ",acc)
prec=precision_score(y_test,y_pred)
print("precision score = ",prec)
rec=recall_score(y_test,y_pred)
print("recall score = ",rec)
f1=f1_score(y_test,y_pred)
print("f1 score = ",f1)
auc=roc_auc_score(y_test,y_pred)
print("roc-auc-score = ",auc)
confuse=confusion_matrix(y_test,y_pred)
print("confusion matrix=\n",confuse)
```
